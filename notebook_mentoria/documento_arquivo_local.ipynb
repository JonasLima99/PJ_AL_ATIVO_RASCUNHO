```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None) 
```

```python
df_principal = pd.read_csv('ALUNOS ATIVOS.csv', encoding = 'utf-8')

df_principal.head(8)
```

```python
df_principal.columns
```

Pretendo seguir a mesma linha de raciocinio que sigo no tratamento via excel, para isso, tenho essas colunas disponíveis:

'ALUNO', 'NOME_ALUNO', 'NOME_SOCIAL', 'PESSOA', 'CPF', 'CURSO','NOME_CURSO' 'CURRICULO', 'SERIE', 'TURMA_PREF', 'DISCIPLINA' ,'NOME_DISCIPLINA', 'TURMA, 'SIT_MATRICULA'.

Preciso saber quais os tipos de dados e se necessário realizar as conversões de dados.

```python
df_principal.dtypes
```

```python
df_tratamento = df_principal[['ALUNO', 'NOME_ALUNO', 'NOME_SOCIAL', 'PESSOA', 'CPF', 'CURSO','NOME_CURSO', 'CURRICULO', 'SERIE', 'TURMA_PREF', 'DISCIPLINA' ,'NOME_DISCIPLINA', 'TURMA', 'SIT_MATRICULA']]
df_tratamento.head(20)
```

Um Aluno ativo, é um aluno matriculado ou trancado, então irei filtrar apenas esses estudantes.

```python
df_filtro_ativo = df_principal[ (df_principal['SIT_MATRICULA'] == 'MATRICULADO') | (df_principal['SIT_MATRICULA'] == 'TRANCADO') ]
df_filtro_ativo.head(30)
```

```python
df_alunos = df_filtro_ativo.drop_duplicates(subset=['CPF'], keep='first')
df_alunos.head(8)
```

Com as duplicadas removidas, filtrei pela coluna do CPF, e o drop só conta no máximo 1 CPF, o restante ele trata como repetida.

Agora, para contagem isso, vou seguir inicialmente com a turma de entrada, que é a 'TURMA_PREF'.

```python
relatorio_final = df_alunos.groupby(['NOME_CURSO', 'CURRICULO', 'TURMA_PREF']).size().reset_index(name='Total_Alunos_Ativos')
relatorio_final.head()
```

```python
relatorio_final.to_csv('Total_Alunos_Ativos.csv', index=False, encoding='utf-8')
```
