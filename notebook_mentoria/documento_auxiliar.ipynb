```python
#Iniciando primeiros toques com pandas 
import pandas as pd
```

```python
# adicionando dicionario normal 
dicionario = {'Nome': ['Raul', 'João', 'Adriano', 'Nicolas', 'Fred', 'Alberto'], 'Idade': [18,29,33,45,16,22], 'Trabalha?': ['Não', 'Sim', 'Sim', 'Não', 'Sim', 'Não'], 'Estuda?': ['Sim', 'Sim', 'Não', 'Sim', 'Não', 'Não'], 'Renda Mensal': [4200, 1700, 8000, 2860, 4200, 10200]}
print(dicionario)
```

```python
#conversão para data frema(pandas)

df = pd.DataFrame(dicionario)
print(df)
```

```python
#Leitura e Escrita de dados em CSV 
#Se o arquivo já existir, ele sobrescreve.
#Salvando arquivo em csv, sem indexação, para não atrapalhar em tabelas futuras.
df.to_csv('dados_brutos', index = False, encoding = 'utf-8')

#O padrão já bem por si só com indexação natural
df.to_csv('dados_brutos_com_index', encoding= 'utf-8')

```

```python
#ler dados de um arquivo para um novo DataFrame 
df_1 = pd.read_csv('dados_Brutos', encoding = 'utf-8')

#Visualizando primeiras linhas
df_1.head(3)
```

```python
#exemplo de indexação duplicada, devido a naturalidade do pandas.
df_2 = pd.read_csv('dados_Brutos_com_index', encoding = 'utf-8')

df_2.head(4)
```

```python
#removendo coluna duplicada
df_2 = df_2.drop(df_2.columns[0], axis = 1)
df_2.head(4)
```

```python
#Selecionando uma única coluna -- sempre que eu selecionar uma unica coluna, o pandas me retorna uma "series".

nomes = df_2['Nome']
print(f'Selecionando a coluna nomes:\n {nomes}')
type(nomes)
```

```python
#Selecionando uma unica linha da coluna selecionada
nome = df_2.loc[2,'Nome']
print(f'O nome da linha pesquisada é: {nome}')
```

```python
#Selecionando um valor de qualquer coluna, da linha selecionada
idade_Adriano = df_2.iloc[2,1]
print(f'A idade de adriano é: {idade_Adriano}')
```
